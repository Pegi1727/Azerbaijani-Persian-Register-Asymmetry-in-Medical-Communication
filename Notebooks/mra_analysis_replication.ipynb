# Multilingual Register Asymmetry (MRA) Replication Notebook
**Study:** Azerbaijani–Persian Register Asymmetry in Medical Communication
**Sample:** Bilingual Azerbaijani–Persian physicians in Iran ($N=150$)

---
### Methodological Notes
- All measures are **self-reported** perceptions, comfort ratings, and reported code-switching tendencies.
- The analysis is strictly **descriptive** (22 invariant survey features across the sample, 10 non-invariant features). No predictive causal target is defined.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Style setup
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['figure.dpi'] = 300

# Candidate paths for the cleaned dataset
candidate_paths = [
    '../data/physician_dataset_cleaned.xlsx',
    'physician_dataset_cleaned.xlsx',
    'physician_dataset_150_model_ready.xlsx',
    '/mnt/data/physician_dataset_cleaned.xlsx',
    '/mnt/data/physician_dataset_150_model_ready.xlsx'
]

data_path = next((p for p in candidate_paths if os.path.exists(p)), None)
if data_path is None:
    raise FileNotFoundError('No valid dataset file found. Please provide physician_dataset_cleaned.xlsx.')

print(f'Loading data from: {data_path}')
excel_obj = pd.ExcelFile(data_path)
sheet = 'Cleaned_Data' if 'Cleaned_Data' in excel_obj.sheet_names else excel_obj.sheet_names[0]
df = pd.read_excel(data_path, sheet_name=sheet)
print(f'Dataset loaded successfully: {df.shape[0]} rows, {df.shape[1]} columns.')
```

## 1. Demographic and Professional Profile
Corresponds to Section 4.1 & Table 1 in the manuscript ($N=150$).

```python
demo_cols = ['Gender', 'Age_Group', 'Experience_Category', 'Specialty']

for col in demo_cols:
    if col in df.columns:
        counts = df[col].value_counts(dropna=False)
        pcts = (df[col].value_counts(normalize=True, dropna=False) * 100).round(1)
        summary = pd.DataFrame({'Count (n)': counts, 'Percentage (%)': pcts})
        print(f'=== Variable: {col} ===')
        print(summary)
        print()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

if 'Specialty' in df.columns:
    df['Specialty'].value_counts().plot(kind='barh', ax=axes[0], color='#2b5c8f', edgecolor='black')
    axes[0].set_title('Specialty Distribution (N=150)', fontsize=12, fontweight='bold')
    axes[0].set_xlabel('Count')
    axes[0].invert_yaxis()

if 'Experience_Category' in df.columns:
    df['Experience_Category'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%',
                                                  colors=['#4a90e2', '#70a1ff'], startangle=140,
                                                  wedgeprops=dict(edgecolor='black'))
    axes[1].set_title('Clinical Experience Categories', fontsize=12, fontweight='bold')
    axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

## 2. Self-Reported Register Access & Lexical Retrieval
Corresponds to Section 4.2, Table 2 & Figure 1.

```python
reg_items = ['Survey_Item_03', 'Survey_Item_04', 'Survey_Item_07']
reg_summary = []

for item in reg_items:
    if item in df.columns:
        counts = df[item].value_counts()
        pcts = (df[item].value_counts(normalize=True) * 100).round(1)
        reg_summary.append({
            'Item': item,
            'Yes (n)': counts.get('Yes', 0),
            'Yes (%)': pcts.get('Yes', 0.0),
            'No (n)': counts.get('No', 0),
            'No (%)': pcts.get('No', 0.0)
        })

pd.DataFrame(reg_summary)
```

## 3. Reported Drivers of Persian Insertion
Corresponds to Section 4.3 & Table 3.

```python
driver_cols = ['Survey_Item_24', 'Survey_Item_25', 'Survey_Item_26']

for col in driver_cols:
    if col in df.columns:
        print(f'=== Item: {col} ===')
        counts = df[col].value_counts(dropna=False)
        pcts = (df[col].value_counts(normalize=True, dropna=False) * 100).round(1)
        print(pd.DataFrame({'Count (n)': counts, 'Percentage (%)': pcts}))
        print()
```
